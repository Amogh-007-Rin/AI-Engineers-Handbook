# Responsible AI slice audit
Measure group outcomes and reject inadequate evidence.

```python
def group_accuracy(rows, minimum=2):
    groups = {}
    for group, expected, predicted in rows:
        correct, total = groups.get(group, (0, 0))
        groups[group] = (correct + (expected == predicted), total + 1)
    if any(total < minimum for _, total in groups.values()):
        raise ValueError('insufficient group evidence')
    return {g: c / n for g, (c, n) in groups.items()}
```

```python
scores = group_accuracy([('a',1,1),('a',0,0),('b',1,0),('b',0,0)])
assert scores == {'a': 1.0, 'b': .5}
try:
    group_accuracy([('a',1,1),('b',1,1)])
except ValueError:
    pass
else:
    raise AssertionError('small slices must fail')
```
